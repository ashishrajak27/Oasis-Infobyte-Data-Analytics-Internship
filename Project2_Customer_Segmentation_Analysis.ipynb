```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# 1. Re-create the Data (Ensuring it's fresh in memory)
data_dict = {
    'Income': [58138, 46344, 71613, 26646, 58293, 62513, 55633, 33454, 30351, 5648, 7500, 44794],
    'MntTotal': [1617, 27, 776, 53, 422, 716, 590, 169, 46, 49, 61, 10]
}
df = pd.DataFrame(data_dict)

# 2. Scale the data (Fast execution)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df[['Income', 'MntTotal']])

# 3. K-Means Clustering (Using n_init to prevent buffering)
# We choose 3 clusters as per the project logic
kmeans = KMeans(n_clusters=3, init='k-means++', n_init=10, random_state=42)
df['Cluster'] = kmeans.fit_predict(X_scaled)

# 4. Immediate Visualization
plt.figure(figsize=(8, 5))
sns.scatterplot(x=df['Income'], y=df['MntTotal'], hue=df['Cluster'], palette='bright', s=100)
plt.title('Project 2: Customer Segments')
plt.grid(True)
plt.show()

print("Execution Successful: Clusters created.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Creating the data based on your project file structure
data_dict = {
    'Income': [58138, 46344, 71613, 26646, 58293, 62513, 55633, 33454, 30351, 5648, 7500, 44794],
    'MntTotal': [1617, 27, 776, 53, 422, 716, 590, 169, 46, 49, 61, 10]
}
df = pd.DataFrame(data_dict)
```

```python
# Scaling is necessary for K-Means to weigh Income and Spending equally
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df[['Income', 'MntTotal']])
```

```python
# Applying K-Means with the optimal 3 clusters
kmeans = KMeans(n_clusters=3, init='k-means++', n_init=10, random_state=42)
df['Cluster'] = kmeans.fit_predict(X_scaled)

# Creating the final plot
plt.figure(figsize=(10, 6))
sns.scatterplot(x=df['Income'], y=df['MntTotal'], hue=df['Cluster'], palette='viridis', s=100)
plt.title('Customer Segmentation: Income vs Spending')
plt.xlabel('Annual Income')
plt.ylabel('Total Spending')
plt.legend(title='Cluster')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
