```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1: Load the data
df = pd.read_csv('Twitter_Data.csv')

# Step 2: Data Cleaning (Removing Nulls)
# Removing the 4 missing text rows and 7 missing category rows
df.dropna(inplace=True)

print("Data Loaded and Cleaned.")
print(f"Total rows after cleaning: {len(df)}")
```

```python
# Step 3: Mapping numeric categories to text labels
# We create a dictionary to map the values
sentiment_map = {-1.0: 'Negative', 0.0: 'Neutral', 1.0: 'Positive'}
df['Sentiment_Label'] = df['category'].map(sentiment_map)

# Step 4: Sentiment Distribution (Frequency Analysis)
sentiment_counts = df['Sentiment_Label'].value_counts()

print("--- Sentiment Distribution ---")
print(sentiment_counts)

# Let's see the percentage distribution as well
print("\n--- Percentage Distribution ---")
print(df['Sentiment_Label'].value_counts(normalize=True) * 100)
```

```python
# Step 5: Visualizing the results

# Setting the visual style
sns.set(style="whitegrid")
plt.figure(figsize=(12, 5))

# 1. Bar Chart
plt.subplot(1, 2, 1)
sns.countplot(data=df, x='Sentiment_Label', palette='viridis', order=['Positive', 'Neutral', 'Negative'])
plt.title('Frequency of Sentiments')
plt.xlabel('Sentiment')
plt.ylabel('Number of Tweets')

# 2. Pie Chart
plt.subplot(1, 2, 2)
df['Sentiment_Label'].value_counts().plot.pie(autopct='%1.1f%%', colors=['#440154', '#21918c', '#fde725'], startangle=140)
plt.title('Sentiment Share Percentage')
plt.ylabel('') # Removes the vertical label for cleaner look

plt.tight_layout()
plt.show()
```
