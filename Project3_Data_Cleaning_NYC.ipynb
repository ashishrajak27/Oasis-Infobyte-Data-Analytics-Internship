```python
import pandas as pd

# Step 1: Load the data
df = pd.read_csv('AB_NYC_2019.csv')

# Step 2: Check for missing values (The Audit)
print("--- Missing Values Per Column ---")
print(df.isnull().sum())

# Look at the data types
print("\n--- Column Data Types ---")
print(df.dtypes)
```

```python
# Step 3: Dropping unnecessary columns
# We remove 'last_review' because it's a date field with many nulls that we don't need for basic analysis
df.drop(['id', 'host_name', 'last_review'], axis=1, inplace=True)

# Step 4: Handling missing values
# If 'reviews_per_month' is NaN, it likely means there were 0 reviews.
df['reviews_per_month'].fillna(0, inplace=True)

# For the few missing 'name' entries, we label them as Unknown
df['name'].fillna("Unknown", inplace=True)

# Verify that the cleaning worked
print("--- Missing Values After Cleaning ---")
print(df.isnull().sum())

print("\n--- Remaining Columns ---")
print(df.columns)
```

```python
# Step 5: Checking for duplicates
duplicate_count = df.duplicated().sum()
print(f"Number of duplicate rows found: {duplicate_count}")

# If duplicates were found, we would remove them:
if duplicate_count > 0:
    df.drop_duplicates(inplace=True)
    print("Duplicates removed.")

# Step 6: Consistency Check (Price)
# Airbnb rooms usually shouldn't have a price of 0. Let's check.
zero_prices = (df['price'] == 0).sum()
print(f"Number of listings with $0 price: {zero_prices}")

# We will remove rows where price is 0 to keep the data realistic
df = df[df['price'] > 0]

print("\n--- Final Cleaning Summary ---")
print(f"Final Row Count: {len(df)}")
df.describe() # This shows a summary of numbers to check for outliers
```

```python
# Step 7: Export the cleaned data
# 'index=False' prevents pandas from adding an extra column for the row numbers
df.to_csv('Cleaned_NYC_Airbnb_2019.csv', index=False)

print("Project 3 Complete!")
print("The file 'Cleaned_NYC_Airbnb_2019.csv' has been saved to your folder.")
```
